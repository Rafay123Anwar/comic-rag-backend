````python
import base64
from mistralai.client import Mistral


client = Mistral(
    api_key="your_mistral_api_key_here"
)


def comic_to_text(image_path: str) -> str:

    with open(image_path, "rb") as image_file:
        image_base64 = base64.b64encode(
            image_file.read()
        ).decode("utf-8")

    prompt = """
    Analyze this comic image carefully.

    Your task is to convert the comic into useful text.

    1. TEXT EXTRACTION
    Extract ALL visible written text:
    - Speech bubbles
    - Thought bubbles
    - Narration boxes
    - Captions
    - Signs
    - Labels
    - Sound effects
    - Any other visible text

    Preserve the wording as accurately as possible.

    2. VISUAL DESCRIPTION
    Describe what is visually happening:
    - Characters
    - Character actions
    - Environment
    - Background
    - Important objects
    - Important events
    - Relationships between characters

    Do not invent anything that is not visible.

    Return exactly this format:

    TEXT:
    ...

    VISUAL DESCRIPTION:
    ...
    """

    response = client.chat.complete(
        model="mistral-small-2603",
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": prompt
                    },
                    {
                        "type": "image_url",
                        "image_url": (
                            f"data:image/jpeg;base64,{image_base64}"
                        )
                    }
                ]
            }
        ]
    )

    return response.choices[0].message.content


result = comic_to_text("comic.jpeg")

print(result)
````

````python
import subprocess as sp
import os
import base64
import tempfile
import rarfile

from mistralai.client import Mistral


# -----------------------------
# Mistral Client
# -----------------------------

client = Mistral(
    api_key="your_mistral_api_key_here"
)


# -----------------------------
# Single Comic Page
# -----------------------------

def comic_page_to_text(image_path: str) -> str:
    """
    Ek comic page ko analyze karta hai aur:
    - visible text extract karta hai
    - visual content describe karta hai
    """

    with open(image_path, "rb") as image_file:
        image_base64 = base64.b64encode(
            image_file.read()
        ).decode("utf-8")

    prompt = """
        Analyze this comic page.

        Your job is to extract information that is actually visible
        or directly readable from the image.

        ========================
        1. TEXT
        ========================

        Extract ALL visible written text:

        - Speech bubbles
        - Thought bubbles
        - Narration boxes
        - Captions
        - Signs
        - Labels
        - Sound effects
        - Titles
        - Any other readable text

        Preserve the wording as accurately as possible.

        Do NOT rewrite or summarize the text.

        If text is partially unreadable, write:
        [unclear]

        If there is genuinely no readable text, write:
        [No visible text]


        ========================
        2. VISUAL DESCRIPTION
        ========================

        Describe ONLY what can be directly observed.

        Include:

        - Characters or figures visible
        - Appearance when clearly visible
        - Actions or poses
        - Objects
        - Environment
        - Background
        - Locations/settings when visually obvious
        - Important visual elements
        - Panel composition

        Do NOT guess:

        - Character identity
        - Character relationships
        - Character emotions unless clearly visible
        - Story events that are not directly shown
        - Character intentions
        - Historical/background information
        - Supernatural explanations
        - What happened before or after the scene

        Avoid words such as:
        "likely", "probably", "possibly", "appears to be"
        unless uncertainty is genuinely unavoidable.

        If something cannot be determined from the image,
        say "Not visually determinable."


        ========================
        OUTPUT FORMAT
        ========================

        TEXT:

        [All visible text]


        VISUAL DESCRIPTION:

        Characters/Figures:
        [Directly visible characters or figures]

        Actions/Poses:
        [What they are visibly doing]

        Environment:
        [Visible setting]

        Objects:
        [Important visible objects]

        Background:
        [Important background details]

        Other Visual Details:
        [Any other directly observable details]
        """

    response = client.chat.complete(
        model="mistral-small-2603",
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": prompt
                    },
                    {
                        "type": "image_url",
                        "image_url": (
                            f"data:image/jpeg;base64,{image_base64}"
                        )
                    }
                ]
            }
        ]
    )

    return response.choices[0].message.content
````

````python
import os
import tempfile
import subprocess as sp
import time

from concurrent.futures import ThreadPoolExecutor, as_completed


def cbr_to_text(
    cbr_path: str,
    max_workers: int = 3,
    max_retries: int = 5
) -> list:

    seven_zip = r"C:\Program Files\7-Zip\7z.exe"

    if not os.path.exists(seven_zip):
        raise FileNotFoundError(
            f"7-Zip not found: {seven_zip}"
        )

    if not os.path.exists(cbr_path):
        raise FileNotFoundError(
            f"Comic file not found: {cbr_path}"
        )

    with tempfile.TemporaryDirectory() as temp_dir:

        # -------------------------
        # Extract CBR
        # -------------------------

        print("📦 Extracting comic...")

        sp.run(
            [
                seven_zip,
                "x",
                cbr_path,
                f"-o{temp_dir}",
                "-y"
            ],
            check=True,
            stdout=sp.PIPE,
            stderr=sp.PIPE,
            text=True
        )

        print("✅ Comic extracted.")

        # -------------------------
        # Find pages
        # -------------------------

        image_extensions = (
            ".jpg",
            ".jpeg",
            ".png",
            ".webp"
        )

        pages = []

        for root, dirs, files in os.walk(temp_dir):

            for file in files:

                if file.lower().endswith(
                    image_extensions
                ):
                    pages.append(
                        os.path.join(root, file)
                    )

        pages.sort()

        total_pages = len(pages)

        if total_pages == 0:
            raise ValueError(
                "No comic pages found."
            )

        print(
            f"📚 Total pages: {total_pages}"
        )

        print(
            f"⚡ Workers: {max_workers}"
        )

        # -------------------------
        # Process page with retry
        # -------------------------

        def process_page(page_info):

            page_number, page_path = page_info

            for attempt in range(
                max_retries + 1
            ):

                try:

                    result = comic_page_to_text(
                        page_path
                    )

                    return {
                        "page": page_number,
                        "text": result,
                        "status": "success"
                    }

                except Exception as e:

                    error_message = str(e)

                    # Check rate limit
                    if "429" in error_message:

                        if attempt >= max_retries:

                            return {
                                "page": page_number,
                                "text": "",
                                "status": "error",
                                "error": error_message
                            }

                        # Exponential backoff
                        wait_time = min(
                            2 ** attempt,
                            30
                        )

                        print(
                            f"⏳ Page {page_number}: "
                            f"Rate limit. "
                            f"Retrying in "
                            f"{wait_time}s..."
                        )

                        time.sleep(wait_time)

                    else:

                        return {
                            "page": page_number,
                            "text": "",
                            "status": "error",
                            "error": error_message
                        }

        # -------------------------
        # Parallel processing
        # -------------------------

        results = []

        page_tasks = [
            (index, path)
            for index, path in enumerate(
                pages,
                start=1
            )
        ]

        with ThreadPoolExecutor(
            max_workers=max_workers
        ) as executor:

            futures = {
                executor.submit(
                    process_page,
                    page
                ): page[0]
                for page in page_tasks
            }

            completed = 0

            for future in as_completed(futures):

                page_number = futures[future]

                result = future.result()

                results.append(result)

                completed += 1

                if result["status"] == "success":

                    print(
                        f"✅ [{completed}/{total_pages}] "
                        f"Page {page_number}"
                    )

                else:

                    print(
                        f"❌ [{completed}/{total_pages}] "
                        f"Page {page_number}"
                    )

        # -------------------------
        # Restore page order
        # -------------------------

        results.sort(
            key=lambda x: x["page"]
        )

        successful = sum(
            1
            for r in results
            if r["status"] == "success"
        )

        failed = total_pages - successful

        print("\n========================")
        print("PROCESSING COMPLETE")
        print("========================")

        print(f"Total   : {total_pages}")
        print(f"Success : {successful}")
        print(f"Failed  : {failed}")

        return results
````

````python
import json
import os


def save_comic_result(
    results: list,
    output_path: str = "comic_result.json",
    comic_name: str = None
):
    """
    CBR processing results ko clean JSON file mein save karta hai.
    """

    # Comic name automatically get karo
    if comic_name is None:
        comic_name = os.path.splitext(
            os.path.basename(output_path)
        )[0]

    data = {
        "comic": comic_name,
        "total_pages": len(results),
        "successful_pages": sum(
            1 for page in results
            if page["status"] == "success"
        ),
        "failed_pages": sum(
            1 for page in results
            if page["status"] == "error"
        ),
        "pages": results
    }

    with open(
        output_path,
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            data,
            file,
            ensure_ascii=False,
            indent=4
        )

    print(
        f"✅ Result saved successfully:\n"
        f"{output_path}"
    )

    return data
````

````python
import json
import time

from app.services.rag_preprocessor import (
    comic_to_rag_documents
)

from app.services.rag_chunker import (
    create_rag_chunks
)

from app.services.embedding import (
    create_embeddings
)

from app.services.vector_store import (
    add_chunks
)


def ingest_comic_to_rag(
    comic_json_path: str
):

    start_time = time.time()

    # -----------------------------
    # Load comic.json
    # -----------------------------

    print("\n[1/5] Loading comic.json...")

    with open(
        comic_json_path,
        "r",
        encoding="utf-8"
    ) as file:

        comic_data = json.load(file)

    print(
        f"   Done: {time.time() - start_time:.2f}s"
    )

    # -----------------------------
    # Documents
    # -----------------------------

    step_start = time.time()

    print("\n[2/5] Creating RAG documents...")

    documents = comic_to_rag_documents(
        comic_data
    )

    print(
        f"   Documents: {len(documents)}"
    )

    print(
        f"   Time: {time.time() - step_start:.2f}s"
    )

    # -----------------------------
    # Chunks
    # -----------------------------

    step_start = time.time()

    print("\n[3/5] Creating chunks...")

    chunks = create_rag_chunks(
        documents,
        chunk_size=800,
        chunk_overlap=150
    )

    print(
        f"   Chunks: {len(chunks)}"
    )

    print(
        f"   Time: {time.time() - step_start:.2f}s"
    )

    if not chunks:

        return {
            "documents": 0,
            "chunks": 0
        }

    # -----------------------------
    # Embeddings
    # -----------------------------

    step_start = time.time()

    print("\n[4/5] Generating embeddings...")

    texts = [
        chunk["content"]
        for chunk in chunks
    ]

    embeddings = create_embeddings(
        texts
    )

    print(
        f"   Embeddings: {len(embeddings)}"
    )

    print(
        f"   Time: {time.time() - step_start:.2f}s"
    )

    # -----------------------------
    # Vector DB
    # -----------------------------

    step_start = time.time()

    print("\n[5/5] Saving to ChromaDB...")

    add_chunks(
        chunks,
        embeddings
    )

    print(
        f"   Time: {time.time() - step_start:.2f}s"
    )

    # -----------------------------
    # Total
    # -----------------------------

    total_time = time.time() - start_time

    print(
        f"\n✅ Ingestion completed in "
        f"{total_time:.2f}s"
    )

    return {
        "documents": len(documents),
        "chunks": len(chunks),
        "embedding_count": len(embeddings),
        "time_seconds": round(
            total_time,
            2
        )
    }
````
